  ==Universal Calculator Template (MAANG Interviews)
  --Big idea:(Explain like interviewer expects)
  ---when reading the expression
  "3 + 2 * 5 - 4"

  --We process it left --> right

 --We keep:
    --num --> current number being built
    --sign --> last operator
    --stack --> stores numbers to combine later.
When we meet an operator:

1. Apply the previous stack
2. Push/update stack
3. Reset number

**At the end --> sum stack


```python
#Universal Python Template

def calculate(s: str) -> int:
    stack = []
    num = 0
    sign = "+"

    #
    for i, c in enumerate(s):

        #if the character is a digit, build the current number
        if c.isdigit():
            num = num * 10 + int(c)

        #if the character is an operator or we are at the end of the string, process the current number 
        if c in "+-*/" or i == len(s) - 1:
            if sign == "+":
                stack.append(num)
            elif sign == "-":
                stack.append(-num)
            elif sign == "*":
                stack[-1] = stack[-1] * num
            elif sign == "/":

                #Python's integer division truncates towards negative infinity, so we need to handle negative numbers correctly
                stack[-1] = int(stack[-1] / num)
            
            #update the sign and reset the current number
            sign = c
            num = 0 

    return sum(stack)

#Example input
s = "3+2*2"

#run the calculator
result = calculate(s)
print(result)
```

```python
# Ultimate MAANG Interview Template (include parentheses and all operators)

#This template solves almost every calculator qeustion
def calculate(s: str) -> int:

    stack = []
    num = 0
    sign = "+"
    result = 0

    #iterate through each character in the string
    for c in s:

        if c.isdigit():
            num = num * 10 + int(c)

        elif c == "(":
            stack.append(result)
            stack.append(sign)
            result = 0
            sign = "+"
            num = 0

        if c in "+-*/)" or c == "":

            if sign == "+":
                result += num
            elif sign == "-":
                result -= num
            elif sign == "*":
                result *= num
            elif sign == "/":
                result = int(result / num)

            num = 0

            if c == ")":
                prev_sign = stack.pop()
                prev_result = stack.pop()

                if prev_sign == "+":
                    result = prev_result + result
                else:
                    result = prev_result - result

            sign = c

    return result




```

```python
# Simple Recursive Parenthesis Calculator Template
def calculate(s: str) -> int:
    
    i = 0
    
    def helper():
        nonlocal i
        stack = []
        num = 0
        sign = "+"
        
        while i < len(s):
            c = s[i]
            
            if c.isdigit():
                num = num * 10 + int(c)
                
            if c == "(":
                i += 1
                num = helper()
                
            if c in "+-*/)" or i == len(s)-1:
                
                if sign == "+": stack.append(num)
                if sign == "-": stack.append(-num)
                if sign == "*": stack.append(stack.pop()*num)
                if sign == "/": stack.append(int(stack.pop()/num))
                
                sign = c
                num = 0
            
            if c == ")":
                return sum(stack)
            
            i += 1
        
        return sum(stack)
    
    return helper()
```
