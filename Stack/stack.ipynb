### Big Idea of Stack
### A stack works like a plate:
 --Push: put the plate on the top
  --Pop: Remove top of the plate
 --Top/Peek: Look at the plate

 Problem: To find the highest value using a stack, we:
 1.Push numbers in the stack
 2. Pop one by one
 3. Keep track of the maximum value seen so far.

 Think like this:
 Stack = [3, 8, 10, 5]
 pop -> 5 -> max = 5
 pop -> 10 -> max = 10
 pop -> 8 -> max = 10
 pop -> 3 -> max = 10
 
 Final answer = 10

```python

class Stack:
     def __init__(self):
          self.stack = []

     def push(self, value):
          self.stack.append(value)

     def pop(self):
          if not self.is_empty():
               return self.stack.pop()
          return None

     def peek(self):
          if not self.is_empty():
               return self.stack[-1]
          return None

     def is_empty(self):
          return len(self.stack) == 0

#Find the Highest value Using Stack
def find_max(stack):
     if stack.is_empty():
          return None
     max_value = stack.pop()
     while not stack.is_empty():
          current_value = stack.pop()
          if current_value > max_value:
               max_value = current_value
     return max_value

# Example:
s = Stack()
s.push(3)
s.push(1)
s.push(4)
s.push(2)
print("Maximum value in the stack:", find_max(s))

#Time and Space Complexity:
# The time complexity of the find_max function is O(n) because it needs to traverse all
# elements in the stack to find the maximum value. The space complexity is O(1) because it uses a constant amount of extra space to store the max_value variable.
#push  - O(1)
#pop   - O(1)
#peek  - O(1)
#is_empty - O(1) 
#find_max - O(n)

```

### ***Interview FOllow-up Questions:***
-Can you design a stack where getMax() returns the maximum element in O(1) time?
-Yes, we can design a stack that keeps track of the maximum element at each level.
-Example: 
-Main stack: [3, 1, 4, 2]
-Max stack:  [3, 3, 4, 4]
-In this design, the getMax() function can simply return the top element of the max
-stack, which will be the maximum element in the main stack at that point in time.
-Then getMax() = top of max stack = O(1) time complexity.
###-The Big idea(Two stacks)
--We use two Stacks:
### --Stack 1 --Normal Stack:
--Stores all values:      [3,8,2,10,5]
### Stack 2 --max Stack
--Stores the maximum values so far
[3, 8, 8, 10, 10]
Meaning:
value pushed       Max so far
3                     3
8                     8
2                     8
10                    10
5                     10
--So the top of max_stack is always the maximum.
### visual example: push sequence
push(3)
stack    =  [3]
max_stack = [3]

push(8)
stack      = [3, 8]
max_stack  = [3, 8]

push(10)
stack      = [3,8,10]
max_stack  = [3,8,8,10]

Now,  getMax() -> 10

```python
#Python Implementation (Production Style)
class MaxStack:
    def __init__(self):
        self.stack = []
        self.max_stack = []  #The top of Max_stack alwasys cotains the maximum value of the stack.

    def push(self, val):
        self.stack.append(val)

        if not self.max_stack:
            self.max_stack.append(val)
        else:
            current_max = self.max_stack[-1]
            self.max_stack.append(max(val, current_max))

    def pop(self):
        if not self.stack:
            return None
        self.max_stack.pop()
        return self.stack.pop()
    def peek(self):
        if not self.stack:
            return None
        return self.stack[-1]
    def get_max(self):
        if not self.max_stack:
            return None
        return self.max_stack[-1]

# Example Usage:      
s = MaxStack()

s.push(3)
s.push(8)
s.push(2)
s.push(10)
s.push(5)

print("Max:", s.get_max())   # 10

s.pop()

print("Max:", s.get_max())   # 10

s.pop()

print("Max:", s.get_max())   # 8

#Time and Space Complexity:

# Operation | Complexity |
# --------- | ---------- |
# push      | O(1)       |
# pop       | O(1)       |
# peek      | O(1)       |
# get_max    | O(1)      |
```
