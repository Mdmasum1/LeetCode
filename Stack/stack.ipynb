### Big Idea of Stack
### A stack works like a plate:
###  --Push: put the plate on the top
###  --Pop: Remove top of the plate
### --Top/Peek: Look at the plate

# Problem: To find the highest value using a stack, we:
# 1.Push numbers in the stack
# 2. Pop one by one
# 3. Keep track of the maximum value seen so far.

### Think like this:
### Stack = [3, 8, 10, 5]
### pop -> 5 -> max = 5
### pop -> 10 -> max = 10
### pop -> 8 -> max = 10
### pop -> 3 -> max = 10
###
### Final answer = 10

```python

class Stack:
     def __init__(self):
          self.stack = []

     def push(self, value):
          self.stack.append(value)

     def pop(self):
          if not self.is_empty():
               return self.stack.pop()
          return None

     def peek(self):
          if not self.is_empty():
               return self.stack[-1]
          return None

     def is_empty(self):
          return len(self.stack) == 0

#Find the Highest value Using Stack
def find_max(stack):
     if stack.is_empty():
          return None
     max_value = stack.pop()
     while not stack.is_empty():
          current_value = stack.pop()
          if current_value > max_value:
               max_value = current_value
     return max_value

# Example:
s = Stack()
s.push(3)
s.push(1)
s.push(4)
s.push(2)
print("Maximum value in the stack:", find_max(s))
```
